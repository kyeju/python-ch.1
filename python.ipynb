```python
#Part2
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""
words = sentence.split()

word_counts = {}

for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print("전체 단어 빈도")
for word, count in word_counts.items():
    print(f"{word} : {count}회")

print("\n2회 이상 등장한 단어")
for word, count in word_counts.items():
    if count >= 2:
        print(f"{word} : {count}회")
```

```python
#Part3
def count_words(sentence):
    words = sentence.split()
    word_counts = {}
    
    for word in words:
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1
            
    return word_counts


def print_frequent_words(word_counts, min_count):
    print(f"\n{min_count}회 이상 등장한 단어")
    for word, count in word_counts.items():
        if count >= min_count:
            print(f"{word} : {count}회")


sentence = """
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

result_counts = count_words(sentence)

print("전체 단어 빈도")
for word, count in result_counts.items():
    print(f"{word} : {count}회")

print_frequent_words(result_counts, 2)
```

```python
#Part4 + 도전문제(특정 단어가 몇 번 등장했는지 검색하기)
def count_words(sentence):
    if not isinstance(sentence, str) or not sentence.strip():
        raise ValueError("입력값이 올바르지 않습니다.")
    
    words = sentence.split()
    word_counts = {}
    
    for word in words:
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1
            
    return word_counts


def search_word(word_counts, target):
    return word_counts.get(target, 0)


sentence = """
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

try:
    result_counts = count_words(sentence)
    
    print("전체 단어 빈도")
    for word, count in result_counts.items():
        print(f"{word} : {count}회")
        
    print("\n" + "=" * 30)
    search_target = input("검색할 단어를 입력하세요: ").strip()
    found_count = search_word(result_counts, search_target)
    
    print(f"'{search_target}' 단어 등장 횟수: {found_count}회")

except ValueError as e:
    print(f"오류 발생: {e}")
except Exception as e:
    print(f"알 수 없는 오류 발생: {e}")
```
